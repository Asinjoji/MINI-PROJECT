```python
import pandas as pd
import schedule
import time
#------------------
#  1.Extract
#------------------
def extract():
  filepath = "/content/drive/MyDrive/DSAB13/Data_Enginnering/salesdata.csv"
  df = pd.read_csv(filepath)
  print("Data extracted sucessfully")
  return df


#------------------------------
#2. Transform
#------------------------------
def transform(df):
  df["total"] = df["qty"] * df["price"]

  print("data transformed sucessfully.")
  return(df)

#------------------------------
# 3.Load
#------------------------------
def load(df):
  output_path = "/content/drive/MyDrive/DSAB13/Data_Enginnering/result.csv"

  df.to_csv(output_path, index=False)
  print("data loaded sucessfully")
  print("Output file:", output_path)


#-------------------------
# 4.Load
#------------------------

def pipeline():
  print("\n---------------------------")
  print("ETL Pipeline started")
  print("---------------------------\n")
  df = extract()
  df = transform(df)
  load(df)


#-------------------------
# 5.Scheduler
#-------------------------
#Run pipeline every 1 minute
schedule.every(10).seconds.do(pipeline)
print("ETL Scheduler Started....")
print("Pipeline will run every 1 minute")
print("Press stop/Interrupt to stop the scheduler")

#Keep checking the schdeule
while True:
  schedule.run_pending()
  time.sleep(1)
```
